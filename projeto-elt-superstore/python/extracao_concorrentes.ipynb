```python
# 1️⃣ Importar bibliotecas
import requests
import pandas as pd
from io import StringIO

# 2️⃣ Definir URL da Wikipédia
url = "https://en.wikipedia.org/wiki/List_of_supermarket_chains"

# 3️⃣ Fazer a requisição HTTP
headers = {"User-Agent": "Mozilla/5.0"}  # Evitar bloqueio
response = requests.get(url, headers=headers)

# 4️⃣ Extrair todas as tabelas HTML usando pandas
dfs = pd.read_html(StringIO(response.text))

# 5️⃣ Selecionar a tabela correta
# Normalmente é a segunda tabela, mas você pode conferir com dfs[i].head()
df_concorrentes = dfs[1]

# 6️⃣ Limpar nomes das colunas (remover espaços e quebras de linha)
df_concorrentes.columns = df_concorrentes.columns.str.strip()

# 7️⃣ Selecionar apenas as colunas relevantes
df_concorrentes = df_concorrentes[
    [
        "Company",
        "Headquarters",
        "Served countries (besides the headquarters)",
        "Number of locations",
        "Number of employees"
    ]
]

# 8️⃣ Remover linhas sem nome da empresa
df_concorrentes = df_concorrentes[df_concorrentes["Company"].notna()]

# 9️⃣ Renomear colunas para padrão snake_case
df_concorrentes = df_concorrentes.rename(columns={
    "Company": "company_name",
    "Headquarters": "headquarters_country",
    "Served countries (besides the headquarters)": "served_countries",
    "Number of locations": "number_of_locations",
    "Number of employees": "number_of_employees"
})

# 🔹 Criar cópia RAW (manter exatamente como extraído)
df_concorrentes_raw = df_concorrentes.copy()

# 10️⃣ Salvar CSV RAW
df_concorrentes_raw.to_csv("competitors_raw.csv", index=False, encoding="utf-8-sig")

# Observações para relatório:
# - A camada RAW mantém exatamente os dados da fonte, incluindo NULLs.
# - Valores NULL aparecem quando a Wikipédia não fornece informação (ex: número de empregados ou mapas).
# - Todas as colunas são strings nesta etapa, sem transformação de tipos.
```
